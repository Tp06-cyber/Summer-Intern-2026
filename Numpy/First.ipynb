```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,30,100)
y=np.sin(x)
plt.plot(x,y, label='sin(x)', color='red', linestyle='dashed')
plt.title=("Sine Wave")
plt.xlabel=("X-axis")
plt.ylabel=("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```
