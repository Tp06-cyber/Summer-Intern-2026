```python
import pandas as pd
import numpy as np
data=[[1,2,3],[3,4,5],[5,6,7]]    

df=pd.DataFrame(data,index=['a','b','c'],columns=['one', 'two', 'three'])
print(df)
df=df.reindex(['x','y','z','a','b','c'])
print(df)
f=df.fillna("-")
print(f)
```

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({
    'one':[1,2,3,4,5],
    'two':[5,4,3,2,1]})
print(df)
df=df.replace({1:100,2:200})
print(df)
```

```python
import pandas as pd
import numpy as np
df=pd.DataFrame({
    'one':['a','b','a','a','b'],
    'two':[5,4,3,2,1]})
print(df)
gd=df.groupby('one')['two'].sum()
print(gd)
g=df.groupby('one')['two'].mean()
print(g)
gde=df.groupby('one')['two'].count()
print(gde)
```

```python
import pandas as pd
data={
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45]
}
df=pd.DataFrame(data)
res=df.sort_values(by='Age',ascending=False)
print(res)
```
