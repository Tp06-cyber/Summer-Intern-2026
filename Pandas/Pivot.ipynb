```python
import pandas as pd
df=pd.DataFrame({
    'col1':range (12),
    'col2':['A']*3+['B']*3+['C']*3+['D']*3,
    'Date':pd.to_datetime(['2026-05-21','2026-05-22','2026-05-23','2026-05-24','2026-05-25','2026-05-26','2026-05-27','2026-05-28','2026-05-29','2026-05-30','2026-05-31','2026-06-01'])
})
print(df)
pivoted=df.pivot(index='Date',columns='col2',values='col1')
print(pivoted)
```

```python
import pandas as pd
data={
    'Course':['CS','Civil','CS','Civil'],
    'Year':[1,2,2,3],
    'Student':[100,150,120,130 ]
}
df=pd.DataFrame(data)
df['Marks']=df['Student'][:3]*10
print(df)
pivoted=pd.pivot_table(df,index='Course',columns='Year',values='Student',aggfunc='sum')
print(pivoted)
```
