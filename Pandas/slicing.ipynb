```python
import pandas as pd
df=pd.DataFrame(['a','b',1,2],['c','d',34,54],['e','f',45,67])
print(df)
df.index=['r1','r2','r3','r4']
res=df.iloc['r1':'r3']
print(res)
res=df.loc['r1':'r3']
print(res)
```

```python
import pandas as pd
df=pd.DataFrame({'A':[1,2,3],'B':[4,5,6]})
print(df)
print("\n Addition\n",df['A']+5)
print("\n Subraction\n",df-5)
print("\n Multiplication\n",df*5)
print("\n Modulus\n",df%2)
```

```python
import pandas as pd
df1=pd.DataFrame({'A':[1,2,3],'B':[4,5,6]})
df2=pd.DataFrame({'A':[11,12,13,43,54],'B':[21,34,5,43,43]},index=[0,1,2,3,4])
print(df1)
print(df2)
print("\n Addition\n",df1+df2)
print("\n Subraction\n",df1-df2)
```

```python
import pandas as pd 
one=pd.DataFrame({
    "Name":['Jayaprakash','Suresh','Ramesh','Abhay'],
    "Subject":['sub1','sub2','sub3','sub4'],
    "Marks":[34,45,56,67]    
},index=[9,8,7,6])
two=pd.DataFrame({
    "Name":['A','B','C','D'],
    "Subject":['sub1','sub2','sub3','sub4'],
    "Marks":[34,45,56,67]    
},index=[9,8,7,6])
res=pd.concat([one,two],keys=['x','y'])
print(res)
tuuu=pd.concat([one,two],ignore_index=True)
print(tuuu)
fuuu=pd.concat([one,two],axis=1)
print(fuuu)
```

```python
import pandas as pd
one =pd.DataFrame({
    "id":[1,2,3,4],
    "Name":['Jayaprakash','Suresh','Ramesh','Abhay'],
    "Subject":['sub1','sub2','sub3','sub4']})
two =pd.DataFrame({ "id":[2,1,3,4],
    "Name":['A','B','C','D'],
    "Subject":['sub1','sub2','sub3','sub4']})
res=pd.merge(one,two,on='id')
print(res)
```

```python
import pandas as pd
left=pd.DataFrame({
    "id":[1,2,3,4],
    "name":["Alice","Bob","Charlie","David"]
})
right=pd.DataFrame({
    "id":[3,4,5,6],
    "age":[25,30,35,40]
}) 
res=left.merge(right,on='id',how='inner')
print(res)  
```
