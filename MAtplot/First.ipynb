```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,30,100)
y=np.sin(x)
plt.plot(x,y, label='sin(x)', color='red', linestyle='dashed')
plt.title("Sine Wave")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.DataFrame({
    'X': [1,2,3,4,5],
    'Y': [2,3,5,7,11]
})
plt.scatter(data['X'], data['Y'], color='green')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
