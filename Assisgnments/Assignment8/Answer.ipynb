```python
import pandas as pd

# Create DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['A', 'B', 'C', 'D']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4, 5],
    'Marks': [80, 90, 85, 75]
})

# Inner Merge
print("INNER MERGE")
print(pd.merge(df1, df2, on='ID', how='inner'))

# Left Join
print("\nLEFT JOIN")
print(pd.merge(df1, df2, on='ID', how='left'))

# Right Join
print("\nRIGHT JOIN")
print(pd.merge(df1, df2, on='ID', how='right'))

# Index Based Join
df1_index = df1.set_index('ID')
df2_index = df2.set_index('ID')

print("\nINDEX BASED JOIN")
print(df1_index.join(df2_index))

# Multiple Keys Merge
df3 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Dept': ['IT', 'HR', 'IT'],
    'Salary': [50000, 40000, 60000]
})

df4 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Dept': ['IT', 'HR', 'IT'],
    'Bonus': [5000, 3000, 7000]
})

print("\nMERGE WITH MULTIPLE KEYS")
print(pd.merge(df3, df4, on=['ID', 'Dept']))
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Amit', 'Ravi']
})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Sumit', 'Ankit']
})

df3 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [80, 90, 85, 95]
})

# Vertical Concatenation
concat_df = pd.concat([df1, df2], ignore_index=True)

print("CONCATENATED DATAFRAME")
print(concat_df)

# Merge with Third DataFrame
result = pd.merge(concat_df, df3, on='ID')

print("\nMERGED DATAFRAME")
print(result)
```
