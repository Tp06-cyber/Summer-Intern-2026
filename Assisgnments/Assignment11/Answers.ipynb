```python
import numpy as np

arr1 = np.array([1, 2, 3])

arr2 = np.array([[4, 5, 6],
                 [7, 8, 9]])

combined = np.vstack((arr1, arr2))

print("Combined Array:")
print(combined)
```

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Array:")
print(arr)

print("\nMaximum Value:")
print(np.max(arr))

print("\nMinimum Value:")
print(np.min(arr))

rows, columns = arr.shape
print("\nRows:", rows)
print("Columns:", columns)

print("\nAll Elements:")
for row in arr:
    for element in row:
        print(element, end=" ")

print("\n\nSpecific Element:")
print(arr[1, 2])

sum_value = 0
for row in arr:
    for element in row:
        sum_value += element

print("\nSum of Elements:")
print(sum_value)

arr1 = np.array([10, 20, 30])
arr2 = np.array([5, 10, 15])

print("\nAddition:")
print(arr1 + arr2)

print("\nSubtraction:")
print(arr1 - arr2)

print("\nMultiplication:")
print(arr1 * arr2)

print("\nDivision:")
print(arr1 / arr2)
```

```python
import numpy as np


arr1 = np.array([[10, 20, 30],
                 [40, 50, 60]])

arr2 = np.array([[15, 25, 35],
                 [45, 55, 65]])

average = (arr1 + arr2) / 2

print("Average of Arrays:")
print(average)

combined = np.concatenate((arr1.flatten(), arr2.flatten()))

print("\nMean:")
print(np.mean(combined))

print("\nMedian:")
print(np.median(combined))

```
