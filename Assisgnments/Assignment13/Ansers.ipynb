```python
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

arr = np.nan_to_num(arr, nan=0)

print("NaN Replaced with 0:")
print(arr)

transpose_arr = arr.T

print("\nRows and Columns Interchanged:")
print(transpose_arr)
```

```python
import numpy as np

arr = np.arange(24).reshape(2, 3, 4)

print("Original Shape:", arr.shape)

new_arr = np.moveaxis(arr, 0, 2)

print("New Shape:", new_arr.shape)
print(new_arr)
```

```python
import numpy as np

arr = np.array([[10, np.nan, 30],
                [40, 50, np.nan],
                [70, 80, 90]])

col_mean = np.nanmean(arr, axis=0)

inds = np.where(np.isnan(arr))

arr[inds] = np.take(col_mean, inds[1])

print(arr)
```

```python
import numpy as np

arr = np.array([10, -5, 20, -8, 30, -2])

arr = np.where(arr < 0, 0, arr)

print(arr)
```

```python
import numpy as np

arr1 = np.array([[10, 20, 30],
                 [40, 50, 60]])

arr2 = np.array([[15, 25, 35],
                 [45, 55, 65]])

average = (arr1 + arr2) / 2

print("Average:")
print(average)

combined = np.concatenate((arr1.flatten(), arr2.flatten()))

print("\nMean:")
print(np.mean(combined))

print("\nMedian:")
print(np.median(combined))

```

```python
import numpy as np

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])

solution = np.linalg.solve(A, B)

print("Using linalg.solve():")
print("x =", solution[0])
print("y =", solution[1])
print("z =", solution[2])

inverse_A = np.linalg.inv(A)

result = np.dot(inverse_A, B)

print("\nUsing Inverse Matrix:")
print("x =", result[0])
print("y =", result[1])
print("z =", result[2])
```
