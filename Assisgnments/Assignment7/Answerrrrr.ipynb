```python
import pandas as pd

# Create Series from Dictionary
dict_data = {'a': 10, 'b': 20, 'c': 30}
s1 = pd.Series(dict_data)
print("Series from Dictionary:")
print(s1)

# Create Series from List
list_data = [100, 200, 300, 400]
s2 = pd.Series(list_data)
print("\nSeries from List:")
print(s2)

# Access Elements of Series
print("\nAccessing Elements:")
print("First Element:", s2[0])
print("Third Element:", s2[2])
print("First Three Elements:")
print(s2[:3])
```

```python
import pandas as pd

# DataFrame from 2D List
data_2d = [[1, 'Ram'], [2, 'Shyam'], [3, 'Mohan']]
df1 = pd.DataFrame(data_2d, columns=['ID', 'Name'])
print("DataFrame from 2D List:")
print(df1)

# DataFrame from Dictionary
data_dict = {
    'Name': ['Ram', 'Shyam', 'Mohan'],
    'Age': [20, 21, 22]
}
df2 = pd.DataFrame(data_dict)
print("\nDataFrame from Dictionary:")
print(df2)

# DataFrame from List of Lists
data_list = [
    [101, 'Amit', 85],
    [102, 'Ravi', 90],
    [103, 'Sumit', 88]
]
df3 = pd.DataFrame(data_list, columns=['Roll', 'Name', 'Marks'])
print("\nDataFrame from List of Lists:")
print(df3)

# DataFrame from List of Tuples
data_tuple = [
    (201, 'Ankit', 75),
    (202, 'Vikas', 80),
    (203, 'Deepak', 85)
]
df4 = pd.DataFrame(data_tuple, columns=['Roll', 'Name', 'Marks'])
print("\nDataFrame from List of Tuples:")
print(df4)

# DataFrame from List of Dictionaries
data_dicts = [
    {'Name': 'Amit', 'Age': 20},
    {'Name': 'Ravi', 'Age': 21},
    {'Name': 'Sumit', 'Age': 22}
]
df5 = pd.DataFrame(data_dicts)
print("\nDataFrame from List of Dictionaries:")
print(df5)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Amit', 'Ravi', 'Sumit'],
    'Marks': [85, 45, 90]
})

# Iterate using iterrows()
print("Using iterrows():")
for index, row in df.iterrows():
    print(index, row['Name'], row['Marks'])

# Iterate using itertuples()
print("\nUsing itertuples():")
for row in df.itertuples():
    print(row.Name, row.Marks)

# Select rows based on condition
print("\nMarks > 50:")
print(df[df['Marks'] > 50])

# Select row using iloc
print("\nSecond Row:")
print(df.iloc[1])

# Limited rows and columns
print("\nSelected Rows and Columns:")
print(df.loc[0:1, ['Name', 'Marks']])

# Drop rows based on condition
print("\nAfter Dropping Marks < 50:")
print(df[df['Marks'] >= 50])

# Insert row at position 1
new_row = pd.DataFrame({'Name': ['Ankit'], 'Marks': [88]})
df_new = pd.concat([df.iloc[:1], new_row, df.iloc[1:]]).reset_index(drop=True)
print("\nAfter Inserting Row:")
print(df_new)

# Convert rows to list
print("\nDataFrame to List:")
print(df.values.tolist())
```
